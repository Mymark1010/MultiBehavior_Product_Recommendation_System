```python
from pathlib import Path

NOTEBOOK_DIR = Path.cwd()
PROJECT_DIR = NOTEBOOK_DIR.parent

RAW_DIR = PROJECT_DIR / "data" / "raw"
EVENTS_PATH = RAW_DIR / "events.csv"

print("Project:", PROJECT_DIR)
print("Raw data:", RAW_DIR)
print("Events path:", EVENTS_PATH)
print("File tồn tại:", EVENTS_PATH.exists())
```

```python
events = pd.read_csv(
    EVENTS_PATH,
    dtype={
        "timestamp": "int64",
        "visitorid": "int32",
        "event": "category",
        "itemid": "int32",
        "transactionid": "float64"
    }
)

print("Đã đọc events.csv thành công")
print("Số dòng, số cột:", events.shape)
```

```python
events["datetime"] = pd.to_datetime(
    events["timestamp"],
    unit="ms",
    utc=True
)

events[
    [
        "timestamp",
        "datetime",
        "visitorid",
        "event",
        "itemid"
    ]
].head()
```

```python
print("Bắt đầu:", events["datetime"].min())
print("Kết thúc:", events["datetime"].max())
print(
    "Số ngày:",
    (
        events["datetime"].max()
        - events["datetime"].min()
    ).days
)
```

```python
print(
    "Số user:",
    events["visitorid"].nunique()
)

print(
    "Số item:",
    events["itemid"].nunique()
)

print(
    "Số giao dịch:",
    events["transactionid"].nunique()
)
```

```python
duplicate_count = events.duplicated().sum()

print(
    "Số dòng trùng hoàn toàn:",
    duplicate_count
)
```
