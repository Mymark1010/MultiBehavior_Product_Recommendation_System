# Kiểm tra nguồn RetailRocket
Notebook chỉ đọc CSV raw; bộ dữ liệu học được tạo duy nhất bởi `scripts/prepare_nextstep.py`. Các số bên dưới được tính từ dữ liệu thật.

```python
from pathlib import Path
import sys
import json
import pandas as pd
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'configs/nextstep.json').is_file())
sys.path.insert(0, str(ROOT))
from IPython.display import display
```

```python
raw = pd.read_csv(ROOT / 'data/raw/events.csv', dtype={'transactionid':'Int64'})
print('Raw events:', len(raw))
display(raw.head())
```

```python
display(raw.dtypes)
display(raw.isna().sum())
print('Exact duplicate rows:', int(raw.duplicated().sum()))
display(raw.event.value_counts())
```

```python
from scripts.prepare_nextstep import clean_events
config = json.loads((ROOT / 'configs/nextstep.json').read_text())
events, cutoffs = clean_events(raw, config)
print('Clean events:', len(events))
print('Cutoffs:', cutoffs)
display(events.groupby('split').size())
```
