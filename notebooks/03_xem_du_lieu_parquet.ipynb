# Xem các bảng Parquet next-step
Chỉ đọc dữ liệu; đổi biến `TABLE` để chọn bảng. Schema và số dòng chuẩn nằm trong manifest.

```python
from pathlib import Path
import sys
import json
import pandas as pd
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'configs/nextstep.json').is_file())
sys.path.insert(0, str(ROOT))
from IPython.display import display
from scripts.dataset import require_complete
folder = require_complete(ROOT / 'data/nextstep')
manifest = json.loads((ROOT / 'reports/nextstep/manifest.json').read_text())
display(pd.DataFrame([{'table':name,'rows':info['rows']} for name,info in manifest['artifacts'].items()]))
```

```python
TABLE = 'nextstep_targets.parquet'
assert TABLE in manifest['artifacts']
frame = pd.read_parquet(folder / TABLE)
display(frame.dtypes)
display(frame.head(10))
```

```python
if 'query_id' in frame:
    selected_query = int(frame.query_id.iloc[0])
    display(frame[frame.query_id.eq(selected_query)])
print('Read only; source Parquet files unchanged.')
```
